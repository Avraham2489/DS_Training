## Part II - Graph Neural Networks (Reading)

```Read and understand what are graph neural networks.```

```Recommended reading:``` [this distill article](https://distill.pub/2021/gnn-intro/).

```You really don't have to read all of it, just make sure you understand the main ideas and the formulation of message passing, as well as some popular GNN architectures such as Graph Convolution Networks (GCN) and Graph Attention Networks (GAT).```

```Here are 4 popular GNN architectures aggregate update formulas:```

1. GCN (ICLR 2017):
$$h_{v_i}^{(l+1)}=\sigma \left( \sum_{v_j \in \mathcal{N}(v_i)}{\frac{1}{c_{ij}} h_{v_j}^{(l)} W^{(l)}} \right)$$
```in simpler words: mean over node neighbors + linear activation + sigmoid```

2. GraphSAGE-max (NIPS 2017):
$$h_{v_i}^{(l+1)}=\max_{v_j \in \mathcal{N}(v_i)}{ \sigma( W^{(l)} h_{v_j}^{(l)} + b^{(l)} )} $$
```in simpler words: linear activation + sigmoid + max over node neighbors```

3. GraphSAGE-sum:
$$h_{v_i}^{(l+1)}=\sum_{v_j \in \mathcal{N}(v_i)}{ \sigma( W^{(l)} h_{v_j}^{(l)} + b^{(l)} )} $$
```in simpler words: linear activation + sigmoid + sum over node neighbors```

4. GAT (ICLR 2018):
$$h_{v_i}^{(l+1)}= \sigma \left( \frac{1}{K} \sum_{k=1}^{K}{ \sum_{v_j \in \mathcal{N}(v_i)}{ \alpha_{ij}^{k} W^{(l)}_{k} h_{v_j}^{(l)} } } \right)  $$
```in simpler words: linear activation + attention over node neighbors + sigmoid```
